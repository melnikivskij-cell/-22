## Завдання 7. Сортування за `datetime` (хронологічно)
Рядки формату `РРРР-ММ-ДД гг:хх:сс` сортуються як текст у правильному хронологічному порядку; для надійності порівнюємо з варіантом через `pd.to_datetime`.

```python
sorted_dt = df.sort_values('datetime')
display(sorted_dt.head(3))
print('Збігається з сортуванням за справжньою датою:',
      sorted_dt.index.equals(df.sort_values('datetime', key=pd.to_datetime).index))
```

## Відповіді на питання для самоперевірки

**1. Що таке DataFrame і Series? Як вони пов'язані?**  
Series — одновимірний масив із мітками (індексом), аналог одного стовпця. DataFrame — двовимірна таблиця, що складається з кількох Series зі спільним індексом; кожен стовпець DataFrame є Series (`df['money']`).

**2. Як створити DataFrame зі словника? Вимоги до довжин?**  
`pd.DataFrame({'col1': [...], 'col2': [...]})` — ключі стають назвами колонок. Усі списки мають бути однакової довжини, інакше виникне `ValueError`.

**3. Методи для початку та кінця таблиці.**  
`head(n)` — перші n рядків, `tail(n)` — останні n рядків (за замовчуванням 5).

**4. Різниця між `info()` та `describe()`.**  
`info()` показує технічну інформацію: назви колонок, кількість непорожніх значень, типи, обсяг пам'яті. `describe()` обчислює статистики (кількість, середнє, std, мінімум, квартилі, максимум) для числових колонок.

**5. Що повертає `df['price'] > 35`? Як використати для фільтрації?**  
Булеву Series (True/False для кожного рядка) — маску. Її передають у квадратні дужки: `df[df['price'] > 35]`.

**6. Чим відрізняються `loc` та `iloc`?**  
`loc` вибирає за мітками індексу, `iloc` — за числовою позицією (з 0). Приклад: після `sorted_df = df.sort_values('money', ascending=False)` вирази `sorted_df.iloc[0]` (перший рядок за порядком) і `sorted_df.loc[0]` (рядок з міткою 0) дають різні рядки.

**7. Як вибрати значення за номером рядка та назвою колонки?**  
`df.loc[5, 'money']` (за міткою рядка) або `df.iloc[5, df.columns.get_loc('money')]` (за позицією).

**8. Метод сортування. Як за двома колонками?**  
`sort_values()`. За двома: `df.sort_values(['cash_type', 'money'])`, також можна `ascending=[True, False]`.

**9. Чи змінює `sort_values()` оригінал? Як зробити, щоб змінив?**  
Ні, за замовчуванням повертає новий DataFrame. Щоб змінити оригінал: `inplace=True` або присвоїти результат `df = df.sort_values(...)`.

**10. Як зберегти DataFrame у CSV без індексу?**  
`df.to_csv('file.csv', index=False)`.

**11. Як поєднати фільтрацію та сортування?**  
Ланцюжком: `df[df['money'] > 35].sort_values('money', ascending=False)`.

**12. Що буде, якщо використати `and` замість `&`?**  
Виникне `ValueError: The truth value of a Series is ambiguous`. Оператор `and` намагається перетворити цілу Series на одне значення True/False, а `&` працює поелементно. Приклад нижче.

```python
try:
    df[(df['money'] > 30) and (df['cash_type'] == 'card')]
except ValueError as e:
    print('ValueError:', e)
ok = df[(df['money'] > 30) & (df['cash_type'] == 'card')]  # правильно
print('Правильний варіант з &: рядків', len(ok))
```
