**Типи даних у датасеті.** Колонки `date`, `datetime`, `cash_type`, `card`, `coffee_name` мають тип `object` (текст), зокрема `datetime` — це рядок, а не справжня дата. ## Крок 5. Індексація: `iloc` та `loc`

```python
row_iloc = df_sales.iloc[2]  # рядок на позиції 2 (третій за рахунком)
print(row_iloc)
```

```python
row_loc = df_sales.loc[2]  # рядок з міткою індексу 2
print(row_loc)
```

```python
value = df_sales.loc[5, 'money']  # значення колонки money у рядку з індексом 5
print(value)
```

Поки індекс стандартний (0, 1, 2 …), `iloc[2]` і `loc[2]` дають один і той самий рядок. Різниця проявиться після сортування (див. крок 7).

## Крок 6. Фільтрація: транзакції із сумою понад 35

```python
mask_high = df_sales['money'] > 35   # булева маска (Series з True/False)
high_sales = df_sales[mask_high]       # лишаються рядки, де маска True
print(high_sales.head())
```

```python
print('Рядків після фільтрації:', len(high_sales), 'із', len(df_sales))
```

## Крок 7. Сортування за сумою (спадання)

```python
sorted_sales = df_sales.sort_values('money', ascending=False)
print(sorted_sales.head())
```

```python
print('iloc[0] — перший рядок у відсортованому порядку:')
print(sorted_sales.iloc[0])
print()
print('loc[0] — рядок з оригінальною міткою 0:')
print(sorted_sales.loc[0])
```

Після сортування мітки індексу залишилися «старими», тому `iloc[0]` і `loc[0]` повертають **різні** рядки: `iloc` бере фізичну позицію, `loc` — мітку.

## Крок 8. Збереження результату (фільтрація + сортування ланцюжком)

```python
final = df_sales[df_sales['money'] > 35].sort_values('money', ascending=False)
final.to_csv(BASE + 'filtered_sales.csv', index=False)  # без колонки індексів
print('Файл збережено!')
```
