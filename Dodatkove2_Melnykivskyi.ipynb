Функція приймає DataFrame та назву категоріальної колонки і повертає новий DataFrame: кожна категорія та кількість її записів, відсортований за спаданням.

```python
import pandas as pd
```

```python
# Підключення Google Диска (у Colab). Поза Colab — використовується локальна папка.
try:
    from google.colab import drive
    drive.mount('/content/drive')
    BASE = '/content/drive/MyDrive/'
except ImportError:
    BASE = ''
    print('Colab не виявлено — використовується локальна папка')
PATH = BASE + 'sales.csv'
print('Шлях до файлу:', PATH)
```

```python
df = pd.read_csv(PATH)
```

### Функція
Використано `value_counts()` (підрахунок записів у групах) та `sort_values()` (сортування за спаданням).

```python
def group_by_count(df, column):
    """Повертає DataFrame: категорії колонки `column` та кількість записів,
    відсортований за кількістю за спаданням."""
    if column not in df.columns:
        raise KeyError(f'Колонки {column!r} немає в DataFrame')
    counts = df[column].value_counts()                      # Series: категорія -> кількість
    counts = counts.sort_values(ascending=False)            # спадання
    return counts.rename_axis(column).reset_index(name='count')
```

### Демонстрація на датасеті продажів (`coffee_name`)

```python
result = group_by_count(df, 'coffee_name')
result
```

### Ще один виклик — для колонки `cash_type`

```python
group_by_count(df, 'cash_type')
```

### Збереження результату у CSV

```python
result.to_csv(BASE + 'coffee_name_counts.csv', index=False)
print('Файл збережено!')
print(pd.read_csv(BASE + 'coffee_name_counts.csv').head(3))
```
